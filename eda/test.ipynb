```python
import numpy as np;

clean_data['Age'] = pd.to_numeric(clean_data['Age'], errors='coerce')
clean_data['Age'] = clean_data['Age'].fillna(clean_data['Age'].mean()).astype(int)

clean_data
```

```python
clean_data['Exp'] = pd.to_numeric(clean_data['Exp'], errors='coerce')
clean_data['Exp'] = clean_data['Exp'].fillna(clean_data['Exp'].mean()).astype(int)
clean_data
```

```python

clean_data['Location'] = clean_data['Location'].replace(4, np.nan)

clean_data['Location'] = clean_data['Location'].fillna(
    clean_data['Location'].mode()[0]
)
clean_data['Location']
```
