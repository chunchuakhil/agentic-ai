```python


from dotenv import load_dotenv
from langchain_openai import ChatOpenAI
```

```python
# Load API key
load_dotenv()

# Initialize OpenAI model
llm = ChatOpenAI(
    model="gpt-6-luna",
    temperature=1
)

# Send prompt
response = llm.invoke("What is Agentic AI? Explain simply.")

# Print response
print(response.content)
```
